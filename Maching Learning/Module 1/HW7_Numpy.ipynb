**1. Створення масивів**

Створіть (кожен пункт оцінюється в 0.5 бала):
- Масив зі списку `[5, 15, 25, 35, 45]`
- Масив з нулів розміром 7
- Масив з одиниць розміром 6

```python
import numpy as np
```

**2. Базові операції**

Створіть Numpy-масив зі значеннями [12, 24, 36, 48, 60].

Виконайте наступні операції над **оригінальним** масивом і запишіть у нову змінну результат (кожен пункт оцінюється в 0.5 бала):
- Помножте всі елементи на 2
- Додайте до всіх елементів 10
- Піднесіть всі елементи до квадрату
- Обчисліть квадратний корінь з усіх елементів

Виведіть результат виконання кожної операції.

```python
array = np.array([12, 24, 36, 48, 60])
print(f"Дано масив: {array}")

doubled = array * 2
added10 = array + 10
squared = array ** 2
sqrt_array = np.sqrt(array)
sqrt_array = np.round(sqrt_array, 2)

print(f"Помножені всі елементи на два: {doubled}")
print(f"Додано до всіх елементів 10: {added10}")
print(f"Піднесено всі елементи до квадрату: {squared}")
print(f"Квадратний корінь з усіх елементів: {sqrt_array}")
```

**3. Статистичні функції**

Створіть Numpy-масив `data` зі значеннями [18, 27, 33, 41, 52, 68, 79].

Обчисліть використовуючи функціонал numpy (кожен пункт оцінюється в 0.5 бала):
- Середнє значення
- Мінімум
- Максимум
- Суму всіх елементів

```python
data = np.array([18, 27, 33, 41, 52, 68, 79])

print(f"Дано масив: {data}")
print(f"Середнє значення масиву: {np.mean(data):.2f}")
print(f"Мінімум масиву: {np.min(data)}")
print(f"Максимум масиву: {np.max(data)}")
print(f"Сума всіх елементів масиву: {np.sum(data)}")
```

**4. Індексація**

Створіть Numpy-масив `arr` зі значеннями [15, 22, 35, 44, 58, 67].

Виведіть (кожен пункт оцінюється в 0.5 бала):
- Перший елемент
- Останній елемент
- Елементи з індексами 1 до 3
- Елементи більші за 40

```python
arr = np.array([15, 22, 35, 44, 58, 67])

first = arr[0]
last = arr[-1]
from_1_to_3 = arr[1:4]        # елементи з індексами 1, 2, 3
greater_40 = arr[arr > 40]    # елементи більші за 40

print(f"Дано масив: {arr}")
print("Перший елемент:", first)
print("Останній елемент:", last)
print("Елементи з індексами 1 до 3:", from_1_to_3)
print("Елементи більші за 40:", greater_40)
```

**5. Двовимірні масиви**

Задано матрицю 3x3.

Виконайте наступні дії з функціоналом Numpy (кожен пункт оцінюється в 0.5 бала):
- Виведіть форму (shape) матриці
- Виведіть перший рядок
- Виведіть другу колонку
- Обчисліть суму всіх елементів
- Обчисліть середнє всіх елементів

```python
matrix = np.array([[2, 4, 6],
                   [8, 10, 12],
                   [14, 16, 18]])
```

```python
print("Форма (shape) матриці:", matrix.shape)
```

```python
print("Перший рядок:", matrix[0,:])
```

```python
print("Друга колонка:", matrix[:, 1])
```

```python
total = np.sum(matrix)
mean_value = np.mean(matrix)

print("Сума всіх елементів:", total)
print("Середнє всіх елементів:", mean_value)
```
