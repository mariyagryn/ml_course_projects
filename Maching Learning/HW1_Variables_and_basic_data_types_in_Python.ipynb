**2. Арифметичні операції**

Створіть дві змінні: `var1` зі значенням 15, `var2` зі значенням 4.

Обчисліть та виведіть (кожен пункт оцінюється в 0.5 бала):
- Суму
- Різницю
- Добуток
- Ділення
- Цілочисельне ділення
- Залишок від ділення
- Піднесення до степеня


```python
var1 = 15
var2 = 4
```

```python
print(f"Сума: {var1 + var2}")                     # сума
print(f"Різниця: {var1 - var2}")                  # різниця
print(f"Добуток: {var1 * var2}")                  # добуток
print(f"Ділення: {var1 / var2}")                  # ділення
print(f"Цілочисельне ділення: {var1 // var2}")    # цілочисельне ділення
print(f"Залишок від ділення: {var1 % var2}")      # залишок від ділення
print(f"Піднесення до степеня: {var1 ** var2}")   # піднесення до степеня
```

```python
last_name = "Hryn"

full_name = f"{name} {last_name}"

print(full_name)           # звичайний запис
print(full_name.upper())   # великими літерами
print(full_name.lower())   # малими літерами
print(len(full_name))      # довжина рядка

print(f"Мене звати {full_name}, мені {age} років.")
```
