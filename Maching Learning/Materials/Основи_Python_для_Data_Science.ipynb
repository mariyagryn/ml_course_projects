
### Кортежі (Tuples) - Незмінні списки
**Навіщо кортежі, якщо є списки?**

Кортежі схожі на списки, але незмінні (immutable). Це дає кілька переваг:

**Коли використовувати кортежі:**
- Для даних, які не повинні змінюватися (coordinates, shape, дати)
- Як ключі словників (списки не можна!)
- Для повернення кількох значень з функції
- Швидші за списки (менше пам'яті, швидша робота)
- Захист від випадкових змін

### Множини (Sets) - Унікальні елементи

**Що таке set?**

Set - це колекція **унікальних** елементів без певного порядку. Автоматично видаляє дублікати!

**Коли використовувати:**
- Потрібні тільки унікальні значення
- Перевірка наявності елемента (дуже швидко!)
- Математичні операції: об'єднання, перетин, різниця


```python
# Створення множини
unique_grades = {85, 90, 78, 92, 88}
empty_set = set()  # Порожня множина
```

```python
# Додавання елементів
fruits = {"яблуко", "банан"}
fruits.add("апельсин")
fruits.add("яблуко")  # Не додасться - вже є!
print(fruits)  # {'яблуко', 'банан', 'апельсин'}
```

```python
# Видалення
fruits.remove("банан")     # Помилка якщо немає
fruits.discard("груша")    # Не викличе помилку
```

```python
# Перевірка наявності (ДУЖЕ швидко!)
print("яблуко" in fruits)  # True
print("груша" in fruits)   # False
```

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

# Об'єднання (union)
print(set1 | set2)  # {1, 2, 3, 4, 5, 6}

# Перетин (intersection)
print(set1 & set2)  # {3, 4}

# Різниця (difference)
print(set1 - set2)  # {1, 2}

# Симетрична різниця
print(set1 ^ set2)  # {1, 2, 5, 6}
```

## Дебагінг коду - Як знайти помилки

Іноді ми хочемо зрозуміти, чому щось працює, або навпаки - ні. Для цього, якщо ми працюємо в Jupyter Notebook, є кілька способів:

### Спосіб 1: Виводити принтами

```python
# Приклад: чому список порожній?
data = []
print(f"DEBUG: data на початку = {data}")

for i in range(-5, 5):
    if i > 10:  # Помилка в логіці!
        data.append(i)
    print(f"DEBUG: ітерація {i}, data = {data}")

print(f"DEBUG: фінальний data = {data}")  # []
```

```python
# Приклад: перевірка типів даних
def process_data(value):
    print(f"DEBUG: value = {value}, type = {type(value)}")

    if isinstance(value, str):
        print("DEBUG: це рядок, конвертуємо в число")
        value = int(value)

    result = value * 2
    print(f"DEBUG: result = {result}")
    return result

process_data("5")
process_data(10)
```

Зазвичай просто додаємо проміжні print і цього достатньо аби виявити помилки.

### Спосіб 2: Використання бібліотеки pdb (Python Debugger)

Для складніших випадків можна використовувати вбудований дебагер:


```python
import pdb

def calculate_average(numbers):
    # Встановлюємо breakpoint - виконання зупиниться тут
    pdb.set_trace()

    total = 0
    for num in numbers:
      total += num
    count = len(numbers)
    average = total / count

    return average

# При виконанні відкриється інтерактивний дебагер
# Команди:
# n (next) - наступний рядок
# c (continue) - продовжити до наступного breakpoint
# p variable - показати значення змінної
# q (quit) - вийти з дебагера

result1 = calculate_average([10, 20, 30, 40])
result2 = calculate_average([101, 201, 301, 401])
print(result1, result2)
```

## Функції - Організація коду

**Що таке функція?**

Функція - це блок коду з назвою, який можна викликати багато разів. Замість того, щоб писати один і той же код знову і знову, ми пишемо його один раз у функції.

**Навіщо потрібні функції:**

1. **Уникнути повторення коду** - написали один раз, використовуємо скрізь
2. **Організація коду** - розбиваємо велику задачу на маленькі частини
3. **Легше читати** - замість 50 рядків коду бачимо `calculate_statistics(data)`
4. **Легше тестувати** - можна перевірити кожну функцію окремо
5. **Легше змінювати** - вносимо зміну в одному місці, а не в 10

```python
# Функція з кількома параметрами та значеннями за замовчуванням
def calculate_price(price, discount=0, tax=0.2):
    discounted_price = price * (1 - discount)
    final_price = discounted_price * (1 + tax)
    return final_price

print(calculate_price(100))              # 120.0
print(calculate_price(100, discount=0.1))  # 108.0
```

```python
# Обчислення статистик
def calculate_statistics(data):
    """
    Обчислює базові статистики для списку чисел
    """
    mean = sum(data) / len(data)
    minimum = min(data)
    maximum = max(data)

    return {
        "mean": mean,
        "min": minimum,
        "max": maximum
    }

grades = [85, 90, 78, 92, 88, 95]
stats = calculate_statistics(grades)
print(stats)
```

```python
# Нормалізація даних
def normalize(value, min_val, max_val):
    """
    Нормалізує значення до діапазону [0, 1]
    """
    return (value - min_val) / (max_val - min_val)

data = [10, 20, 30, 40, 50]
normalized = [normalize(x, min(data), max(data)) for x in data]
print(normalized)  # [0.0, 0.25, 0.5, 0.75, 1.0]
```

```python
# Перевірка якості моделі
def evaluate_model(predictions, reality):
    """
    Обчислює точність класифікації
    """
    correct = sum([1 for p, r in zip(predictions, reality) if p == r])
    accuracy = correct / len(reality)
    return accuracy

y_true = [1, 0, 1, 1, 0, 1]
y_pred = [1, 0, 1, 0, 0, 1]
accuracy = evaluate_model(y_pred, y_true)
print(f"Точність моделі: {accuracy:.2%}")  # 83.33%
```


### Lambda-функції (анонімні функції)

Lambda - це коротка одноразова функція **на один рядок**, яку не треба іменувати. Використовуємо, коли функція проста і потрібна тільки один раз.

**Коли використовувати:**

1. **Коротка логіка** - функція робить щось просте в 1 рядок
2. **Одноразове використання** - не будемо викликати знову
3. **Як аргумент** - передаємо як функцію в функції вищого порядку `map()`, `filter()`, `sorted()`

```python
# ❌ Довго - створюємо повну функцію
def square(x):
    return x ** 2

result = square(5)
print(result)


```

```python
# ✅ Коротко - lambda на один раз
result = (lambda x: x ** 2)(5)
print(result)
```

```python
# Використання з функціями вищого порядку
numbers = [1, 2, 3, 4, 5]

# map - це спосіб застосувати функцію до кожного елемента
squares = list(map(lambda x: x**2, numbers))
print(squares)  # [1, 4, 9, 16, 25]
```

```python
list(map(square, numbers))
```

```python
# filter - відфільтрувати елементи
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)  # [2, 4]
```
