**1. Цикл for - базовий**

Використовуючи цикл `for`, виведіть числа від 1 до 7 включно використовуючи функцію range.


```python
print("Вивід чисел від 1 до 7 включно:")
for marks in range(1,8):
  print(marks)
```


---

**2. Цикл for по списку**

Дано список:

```python
names = ["Олег", "Анна", "Іван", "Марія"]
```


Використовуючи цикл, виведіть кожне ім'я з привітанням: "Привіт, [ім'я]!"


```python
for greetings in names:
  print(f"Привіт, {greetings}!")
```

**3. Цикл з enumerate**

Дано список оцінок:

```python
grades = [85, 92, 78, 95, 88]
```


Використовуючи `enumerate`, виведіть кожну оцінку з номером: "Студент 1: 85 балів"


```python
for i, grade in enumerate(grades):
  print(f"Студент {i+1}: {grade} балів")
```

**4. Сума парних чисел**

Напишіть програму, яка обчислює суму всіх парних чисел від 1 до 20.

```python
print("Парні числа від 1 до 20:")
even_numbers = list(range(2,21,2))
even_numbers
```

```python
sum_even = sum(even_numbers)
print(f"Сума всіх парних чисел від 1 до 20: {sum_even}")
```

```python
# Альтернативні варіанти
print(f"Сума всіх парних чисел від 1 до 20: {sum(range(2, 21, 2))}")

total = sum(x for x in range(2, 21) if x % 2 == 0)
print("Сума всіх парних чисел від 1 до 20:", total)


total = 0

for x in range(2, 21):
    if x % 2 == 0:
        total += x

print("Сума всіх парних чисел від 1 до 20:", total)
```
