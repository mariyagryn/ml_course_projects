# Розділ 2

## Корисні гарячі клавіші

Позначення:  
- ⌘ — Command  
- ⌥ — Option (Alt)  
- ⌃ — Control  
- ⇧ — Shift  

### 1. Базове редагування

- ⌘X — вирізати виділене  
- ⌘C — копіювати  
- ⌘V — вставити  
- ⌘Z — скасувати (Undo)  
- ⇧⌘Z або ⌘Y — повторити (Redo)  
- ⌘A — виділити все  

- ⌘/ — закоментувати / розкоментувати рядок  
- ⌥↑ / ⌥↓ — рухати рядок вгору / вниз  
- ⇧⌥↓ — продублювати рядок вниз  
- ⇧⌥↑ — продублювати рядок вгору  
- ⌘] / ⌘[ — збільшити / зменшити відступ (indent / outdent)  

### 2. Переміщення по тексту

- ⌘← / ⌘→ — в початок / кінець рядка  
- ⌘↑ / ⌘↓ — в початок / кінець файлу  
- ⌥← / ⌥→ — по словах  
- ⌘Backspace — видалити до початку рядка  
- ⌥Backspace — видалити слово ліворуч  

### 3. Мультикурсор і виділення

- ⌥ + click — додати ще один курсор  
- ⌥⌘↑ / ⌥⌘↓ — додати курсор вище / нижче  
- ⌃⇧⌘→ / ⌃⇧⌘← — виділити колонку (column selection)  
- ⌘D — додати наступне входження виділеного слова  
- ⇧⌘L — виділити **всі** входження виділеного слова  

### 4. Пошук і заміна

- ⌘F — знайти в поточному файлі  
- ⌥⌘F — заміна в поточному файлі  
- ⇧⌘F — глобальний пошук по проєкту  
- ⇧⌥⌘F — глобальна заміна  

### 5. Навігація по файлах і символах

- ⌘P — швидко відкрити файл (Quick Open)  
- ⌘B — показати / сховати бокову панель (Explorer)  
- ⇧⌘O — перейти до символу в файлі (функція, клас тощо)  
- F12 — перейти до визначення (Go to Definition)  
- ⌥F12 — Peek Definition (прев’ю визначення)  
- ⌘T — пошук по всіх символах у робочому просторі  

### 6. Керування вкладками та вікнами

- ⌘N — новий файл  
- ⌘O — відкрити файл  
- ⌘S — зберегти  
- ⇧⌘S — зберегти як…  
- ⌥⌘S — зберегти всі  

- ⌘W — закрити вкладку  
- ⇧⌘T — відкрити останню закриту вкладку  
- ⌘\ — розділити редактор (split)  
- ⌘1 / ⌘2 / ⌘3 — фокус у 1-шу / 2-гу / 3-тю групу редакторів  

### 7. Термінал

- ⌃` — показати / сховати вбудований термінал  
- ⌃⇧` — новий термінал  
- ⌘K потім ⌘S — гарячі клавіші (відкрити keybindings)  

### 8. Вигляд (View)

- ⌘J — показати / сховати панель знизу (термінал / output)  
- ⇧⌘E — фокус на Explorer  
- ⇧⌘F — фокус на пошуку  
- ⇧⌘D — фокус на Debug  
- ⇧⌘X — фокус на Extensions  
- ⌃⌘F — повноекранний режим  

### 9. Дебаг (Debug)

- F9 — поставити / прибрати breakpoint  
- F5 — старт / продовження  
- ⇧F5 — зупинити  
- F10 — Step Over  
- F11 — Step Into  
- ⇧F11 — Step Out
